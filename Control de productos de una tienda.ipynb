```python
CAPACIDAD = 5
productos = []


def agregar_producto():
    if len(productos) >= CAPACIDAD:
        print("\nEl vector está lleno. No se pueden agregar más productos.")
        return

    nombre = input("Ingrese el nombre del producto: ")

    productos.append(nombre)
    print(f"Producto '{nombre}' agregado correctamente.")


def mostrar_productos():
    if len(productos) == 0:
        print("\nNo hay productos registrados.")
        return

    print("\n--- PRODUCTOS REGISTRADOS ---")

    for i, producto in enumerate(productos):
        print(f"Posición {i}: {producto}")


def buscar_producto():
    if len(productos) == 0:
        print("\nNo hay productos para buscar.")
        return

    nombre = input("Ingrese el producto que desea buscar: ")

    encontrado = False

    for i, producto in enumerate(productos):
        if producto.lower() == nombre.lower():
            print(f"Producto encontrado en la posición {i}.")
            encontrado = True
            break

    if not encontrado:
        print("El producto no se encuentra registrado.")


def modificar_producto():
    if len(productos) == 0:
        print("\nNo hay productos para modificar.")
        return

    mostrar_productos()

    try:
        posicion = int(input("\nIngrese la posición del producto que desea modificar: "))

        if posicion < 0 or posicion >= len(productos):
            print("La posición no existe.")
            return

        nuevo_nombre = input("Ingrese el nuevo nombre del producto: ")

        productos[posicion] = nuevo_nombre

        print("Producto modificado correctamente.")

    except ValueError:
        print("Debe ingresar un número válido.")


def eliminar_producto():
    if len(productos) == 0:
        print("\nNo hay productos para eliminar.")
        return

    mostrar_productos()

    try:
        posicion = int(input("\nIngrese la posición del producto que desea eliminar: "))

        if posicion < 0 or posicion >= len(productos):
            print("La posición no existe.")
            return

        eliminado = productos.pop(posicion)

        print(f"Producto '{eliminado}' eliminado correctamente.")

    except ValueError:
        print("Debe ingresar un número válido.")


def menu():
    while True:
        print("\n===== CONTROL DE PRODUCTOS =====")
        print("1. Agregar producto")
        print("2. Mostrar productos")
        print("3. Buscar producto")
        print("4. Modificar producto")
        print("5. Eliminar producto")
        print("6. Salir")

        opcion = input("Seleccione una opción: ")

        if opcion == "1":
            agregar_producto()

        elif opcion == "2":
            mostrar_productos()

        elif opcion == "3":
            buscar_producto()

        elif opcion == "4":
            modificar_producto()

        elif opcion == "5":
            eliminar_producto()

        elif opcion == "6":
            print("\nPrograma finalizado.")
            break

        else:
            print("\nOpción no válida. Intente nuevamente.")


menu()
```
