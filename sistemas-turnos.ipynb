```python
class ColaTurnos:
    def __init__(self, capacidad):
        self.capacidad = capacidad
        self.cola = []

    def encolar(self, nombre):
        if len(self.cola) >= self.capacidad:
            print("La cola está llena. No se puede agregar más personas.")
        else:
            self.cola.append(nombre)
            print(f"{nombre} fue agregado a la cola.")

    def desencolar(self):
        if len(self.cola) == 0:
            print("La cola está vacía. No hay personas para atender.")
        else:
            atendido = self.cola.pop(0)
            print(f"Se está atendiendo a: {atendido}")

    def siguiente(self):
        if len(self.cola) == 0:
            print("No hay personas esperando.")
        else:
            print(f"La siguiente persona es: {self.cola[0]}")

    def mostrar(self):
        if len(self.cola) == 0:
            print("La cola está vacía.")
        else:
            print("\nPersonas en espera:")
            for posicion, nombre in enumerate(self.cola, start=1):
                print(f"Turno {posicion}: {nombre}")

cola = ColaTurnos(5)


cola.encolar("Ana")
cola.encolar("Carlos")
cola.encolar("María")

cola.mostrar()

cola.siguiente()

cola.desencolar()

cola.mostrar()

cola.encolar("Pedro")
cola.encolar("Laura")
cola.encolar("Sofía")
cola.encolar("Juan")
cola.desencolar()
cola.desencolar()
cola.desencolar()
cola.desencolar()
cola.desencolar()
cola.desencolar()
```
